```python
"""
1. Hacer un programa en Python que resuelva la siguiente expresión aritmética
del área de un Triángulo, donde el usuario ingresa la base y la altura.

area = (base)(altura) / 2
"""
# Users input request.
while True:
    try:
        base = float(input("Ingresa el valor numérico de la base del triángulo: "))
        height = float(input("Ingresa el valor numérico de la altura del triángulo: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
        

# Triangle area calculation with a lambda function.
area = lambda b, h: (b * h) / 2
# Print area calculation result.
print(f"El área del triángulo con base {base} y altura {height} es: {area(base, height)}")
```

```python
"""
2. Hacer un programa en Python que resuelva la formula general para resolver ecuaciónes de 2o grado.
El usuario ingresa los valores de a, b y c.
"""
# Users input request.
while True:
    try:
        a = float(input("Ingresa el valor numérico de a: "))
        b = float(input("Ingresa el valor numérico de b: "))
        c = float(input("Ingresa el valor numérico de c: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# x1 and x2 calculation with lambdas functions.
x1 = lambda a, b, c: (-b + (((b**2.0) - 4.0 * (a * c))**0.5)) / (2.0 * a)
x2 = lambda a, b, c: (-b - (((b**2.0) - 4.0 * (a * c))**0.5)) / (2.0 * a)
# x1 and x2 calculation with users input.
try:
    result_x1 = x1(a, b, c)
    result_x2 = x2(a, b, c)
except ValueError as e:
    print(e, type(e))
# Print result.
print(f"Los dos valores que resuelven la ecuación con los parámetros: a = {a}, b = {b}, c = {c} son: X1 = {result_x1:.2f} y X2 = {result_x2:.2f}")
```

```python
"""
3. Hacer un programa en Python que calcule el salario semanal de un obrero, según los siguientes criterios:
- Si trabajó 40 horas o menos, se le paga $16 por hora.
- Si trabajó más de 40 horas, se le paga $16 por las primeras 40 horas, y $20.
por cada hora extra.
"""
# Users input request.
while True:
    try:
        hours = float(input("Ingrese el número de horas trabajadas: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Adjust payment according to worked hours.
if hours <= 40.0:
    payment = hours * 16.0
else: 
    payment = (40 * 16) + ((hours - 40) * 20)

# Print result
print(f"El salario a pagar por {hours:.0f} horas trabajadas es: ${payment:.2f}")
```

```python
"""
4. Hacer un programa en Python que clasifique el tipo de triángulo a partir de sus
lados ingresados:
- Equilátero: Los 3 son iguales.
- Isósceles: Tiene 2 lagos iguales y un lado distinto.
- Escaleno: Los 3 lados son distintos.
"""

# Users input request.
while True:
    try:
        equal_sydes = int(input("Ingrese el número de lados que son iguales en el triángulo: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Match a case.
match equal_sydes:
    case 3:
        print(f"Si presenta {equal_sydes} lados iguales es un triángulo equilátero.")
    case 2:
        print(f"Si presenta {equal_sydes} lados iguales es un triángulo isósceles.")
    case 0:
        print(f"Si ningún de sus lados son iguales es un triángulo escaleno.")
    case _:
        print(f"Con {equal_sydes} lados iguales no tienes ningún tipo de triángulo!.")
```

```python
"""
6. Hacer un programa en Python que, por un número entero dado por el usuario,
muestre la lista de números pares inferiores a él.
"""
# Users input request.
while True:
    try:
        number = int(input("Ingresa un numero y observa los numeros pares inferiores a este: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Initialize an empty even numbers list
even_numbers = []
for n in range(number):
    if n % 2 == 0:
        if n == 0:
            continue
        else:
            even_numbers.append(n)
# Sort the list in reverse
even_numbers.sort(reverse=True)
# Print the result
print(f" Los valores pares por debajo del número {number} son: ")
for e in even_numbers:
    print(e)
```

```python
"""
10. Hacer un programa en Python que imprima la serie y además el valor obtenido de la sumatoria Σn2 , 
donde n es ingresado por el usuario (n no debe ser mayor de 5).
"""
# Users input request.
while True:
    try:
        number = int(input("Ingresa un número y obten la suma de 1 /n : "))
        # Check number value <= 5.
        if number > 5:
            print("El valor numéricos a ingresar debe ser menor a 5")
            continue
        # Break the while loop.
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos menores a 5.")
result = 0
for n in range(1, number + 1):
    print(f"==> {n}**2")
    result += n**2
print(f"La sumatoria de todos los valores de n**1 ... n**{number} es: {result}")
```

```python
"""
11. Hacer un programa en Python que permita registrar las calificaciones de varios estudiantes. 
Cada estudiante tiene un nombre y una lista de calificaciones, implementa funciones para:
a. Agregar un estudiante
b. Agregar una calificación
c. Calcular el promedio de cada estudiante
"""
# Initialize the students dictionary.
students = {}
#  A function dedicated to add a student and an empty list of score values.
def add_student():
    while True:
        name = input("Ingresa el nombre de un estudiante para agregarlo al listado: ").strip()
        # Check if name is text.
        if name.replace(" ", "").isalpha():
            students[name] = []
            break
        else:
            print("Ingresa un nombre valido")
# A function dedicated to add a students score.
def add_score():
    while True:
        name = input("Ingresa el nombre del estudiante del que desea agregar una calificación: ").strip()
        # Check if name is in the dictionary students.
        if name in students:
            break
        else:
            print("El nombre del estudiante no se encuentra en el listado")
    while True:
        try: 
            score = float(input("Ahora ingresa el valor numérico de la calificación"))
            # Add the given score to the list associated to the student.
            students[name].append(score)
            break
        except ValueError as e:
            print(e, type(e))
            print("Debes ingresar solo valores numéricos.")
# A simple function to calculate the averange of each students score.
def averange_calculation(students_dict):
    # For every key an value in the given dictionary.
    for k, v in students_dict.items():
        # Print the name of the student and the averange calculation.
        print(f"{k} --> promedio de calificaciónes: {sum(v)/len(v)}")
# Call add_students function.
add_student()
# Call add_score function.
add_score()
# Call averange_calculation function.
averange_calculation(students)
```

```python
"""
12. Crea un programa en Python para gestionar un pequeño catálogo de productos. Cada producto tendrá:
a. un código clave
b. una tupla con nombre, precio y cantidad disponible
Implementa funciones para:
a. Agregar productos.
b. Consultar un producto por código.
c. Calcular el valor total del inventario.
"""
# Initialize the students dictionary.
products = {'A001': ('Camisa', 12.3, 45)}
#  A simple function dedicated to add a product and its elements.
def add_products():
    while True:
        code = input("Ingresa el código del producto para agregarlo al catálogo: ").strip()
        # Create the product key in products dictionary.
        products[code] = 0
        # Product name input.
        product_name = input("Ingresa el nombre del producto: ").strip()
        try:
            # Price and quantity input.
            price = float(input("Ingresa el precio del producto: "))
            quantity = int(input("Ingresa la cantidad disponible del producto: "))
            # Add the elements of the product into a tuple.
            products[code] = (product_name, price, quantity)
            break
        except ValueError as e:
            print(e, type(e))
            print("Debes ingresar solo valores numéricos.")
# A simple funtion to consult a product in the catalog.
def consult_product():
    while True:
        code = input("Ingresa el código del producto que desea consultar en inventario: ").strip()
        if code in products:
            print(f"El producto con código: {code} es: {products[code][0]}, con un valor por pieza de: {products[code][1]} con un inventario de: {products[code][2]} piezas")
            break
        else:
            print("Ingrese el código de un producto que se encuentre en el catálogo")
# A simple function to calculate the toal value of the product.
def inventory_value():
    while True:
        code = input("Ingresa el código del producto del que desea consultar el valor en inventario: ").strip()
        if code in products:
            print(f"El producto con código: {code} es: {products[code][0]}, con un valor total de: ${products[code][1] * products[code][2]} para {products[code][2]} piezas.")
            break
        else:
            print("Ingrese el código de un producto que se encuentre en el catálogo")
# Call add_products function.
add_products()
# Call consult_product function.
consult_product()
# Call consult_product function.
inventory_value()
print(products)
```

```python
"""
13. Escribe una función en Python que reciba una frase y devuelva un diccionario con la cantidad de veces 
que aparece cada palabra. Ignora mayúsculas y signos de puntuación.
"""
# Users input.
while True:
        text = input("Ingrese una frase o texto corto: ").strip().lower()
        break
# Split the text using split and space character.
words = text.split(" ")
# Create the dictionary using each word as a key and .count() method to assign the times the word is present in list words.
words_count = {word: words.count(word) for word in words}
# Print key value relation in words_count.
for k, v in words_count.items():
        print(f"-{k}---> {v}")
```

```python
"""
14. Crea un programa en Python para gestionar una agenda donde cada contacto tiene:
    a. un nombre (clave del diccionario).
    b. una tupla con teléfono y correo.
Implementa funciones para:
    a. Agregar un contacto.
    b. Buscar un contacto por nombre.
    c. Actualizar datos de un contacto.
"""
# Initialize the agenda dictionary.
agenda = {'Maria': ('55131426', 'maria@gmail.com')}
#  A simple function dedicated to add a product and its elements.
def add_products():
    while True:
        name = input("Ingresa el nombre del contacto: ").strip()
        # Create the contact key in agenda dictionary.
        agenda[name] = 0
        # Contact phone number and email input.
        c_phone_number = input("Ingresa el número telefonico del contacto: ").strip()
        c_email = input("Ingresa el correo electronico del contacto: ").strip()
        agenda[name] =(c_phone_number, c_email)
        break
# A simple funtion to consult a contact in the agenda.
def consult_contact():
    while True:
        name = input("Ingresa el nombre del contacto que desea consultar en la agenda: ").strip()
        if name in agenda:
            print(f"El contacto: {name} tiene el número telefonico: {agenda[name][0]} y el email: {agenda[name][1]}")
            break
        else:
            print("Ingrese un nombre de contacto que se encuentre en la agenda")
# A simple funtion to update a contact data.
def update_contact():
    while True:
        name = input("Ingresa el nombre del contacto que desea actualizar en la agenda: ").strip()
        if name in agenda:
            new_c_phone_num = input("Ingresa el nuevo número telefonico del contacto: ").strip()
            new_c_email = input("Ingresa el nuevo correo electronico del contacto: ").strip()
            agenda[name] =(new_c_phone_num, new_c_email)
            print(f"El contacto: {name} ahora tiene el número telefonico: {agenda[name][0]} y el email: {agenda[name][1]}")
            break
        else:
            print("Ingrese un nombre de contacto que se encuentre en la agenda")

# Call add_products() function.
add_products()
# Call consult_contact() function.
consult_contact()
# Call update_contact() function.
update_contact()
print(agenda)
```

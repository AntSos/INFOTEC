```python
x = sp.symbols('x', real=True)
c = sp.symbols('c', real=True)                          
n = sp.symbols('n', real=True)                          
a_base = sp.symbols('a', positive=True, real=True)      
coefs = sp.symbols('a0:4')                              

funciones_licenciatura = {
    # --- CONSTANTES Y ALGEBRAICAS BÁSICAS ---
    "Constante pura": c,
    "Identidad": x,
    "Potencia cuadrada": x**2,
    "Potencia enésima (General)": x**n,
    "Raíz cuadrada": sp.sqrt(x),
    "Polinomio cúbico general": coefs[3]*x**3 + coefs[2]*x**2 + coefs[1]*x + coefs[0],
    "Valor absoluto": sp.Abs(x),
    
    # --- EXPONENCIALES Y LOGARÍTMICAS ---
    "Constante de Euler (e^1)": sp.exp(1),
    "Constante e^2": sp.exp(2),
    "Exponencial natural (e^x)": sp.exp(x),
    "Exponencial compuesta (e^(x^2))": sp.exp(x**2),
    "Exponencial de base general (a^x)": a_base**x,
    "Logaritmo natural (ln(x))": sp.ln(x),
    "Logaritmo base cualquiera (log_a(x))": sp.log(x, a_base),
    
    # --- TRIGONOMÉTRICAS DIRECTAS ---
    "Seno": sp.sin(x),
    "Coseno": sp.cos(x),
    "Tangente": sp.tan(x),
    "Secante": sp.sec(x),
    "Cosecante": sp.csc(x),
    "Cotangente": sp.cot(x),
    
    # --- TRIGONOMÉTRICAS INVERSAS ---
    "Arcoseno": sp.asin(x),
    "Arcocoseno": sp.acos(x),
    "Arcotangente": sp.atan(x),
    "Arcosecante": sp.asec(x),
    "Arcocosecante": sp.acsc(x),
    "Arcocotangente": sp.acot(x),
    
    # --- HIPERBÓLICAS DIRECTAS E INVERSAS ---
    "Seno hiperbólico": sp.sinh(x),
    "Coseno hiperbólico": sp.cosh(x),
    "Tangente hiperbólica": sp.tanh(x),
    "Argumento seno hiperbólico (asinh)": sp.asinh(x),
    "Argumento coseno hiperbólico (acosh)": sp.acosh(x),
    "Argumento tangente hiperbólica (atanh)": sp.atanh(x),
}

for nombre, f in funciones_licenciatura.items():
    df = sp.diff(f, x)
    print(nombre)
    display(sp.Eq(sp.Derivative(f, x), df))
    print()
```

```python
f5 = x**3 + 3*x**2
display(f5)
f5_prima = sp.diff(f5, x)
display(f5_prima)
# Puntos criticos.
puntos_criticos = sp.solve(f5_prima, x)
display(puntos_criticos)

for p in puntos_criticos:
    evaluacion = sp.diff(f5, x, 2).subs(x, p)

    if evaluacion > 0:
        print(f"Minimo en: f({p}, {f5.subs(x, p)})")

    elif evaluacion < 0: 
        print(f"Maximo en: f({p}, {f5.subs(x, p)})")

    else: 
        print(f"La prueba no es concluyente en: f({p}, {f5.subs(x, p)})")
```

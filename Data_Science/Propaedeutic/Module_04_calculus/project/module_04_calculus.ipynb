```python
import sympy as sp 
import numpy as np 
import pandas as pd 
```

```python
# Global symbolic variables
global x, y, z, a
x, y, z, a = sp.symbols('x y z a', real=True)
```

```python
def delta_function(function):
    """A simple function to calculate the delta of a function"""
    # Function delta as a matrix.
    return sp.Matrix([sp.diff(function, x), sp.diff(function, y)])
```

```python
def solve_d_sys_cero(function):
    """A simple function to solve the delta system equal to 0"""
    # Gradient.
    fx = sp.diff(function, x)
    fy = sp.diff(function, y)
    # Resolve the delta sysyem equal to 0.
    critical_points = sp.solve((fx, fy), (x, y), dict=True)
    # Filter real points in list comprenhension, to avoid I numbers.
    critical_points = [point for point in critical_points if all(value.is_real for value in point.values())]
    return critical_points
```

```python
def hess_matrix_det(function):
    """A simple function dedicated to calculating the Hessian matrix and its determinant"""
    # Using .hessian() method.
    H = sp.hessian(function, (x, y))
    # H matrix determinant using .det() method.
    D = sp.simplify(H.det())
    return H, D
```

```python
def classify_critical_points(function):
    """A simple function dedicated to classify critical points (p₀): if D>0 y fxx>0 → minimum; D>0 y fxx<0 → maximun; D<0 → silla; D=0 →
inconclusive."""
    # Derivate.
    fx = sp.diff(function, x)
    fy = sp.diff(function, y)

    critical_points = sp.solve((fx, fy), (x, y), dict=True)
    # Filter real points in list comprenhension, to avoid I numbers.
    critical_points = [point for point in critical_points if all(value.is_real for value in point.values())]
    # Hessian matrix 
    H = sp.hessian(function, (x, y))
    # Results.
    results = []

    for point in critical_points:
        print(point)
        # Determinant.
        D = sp.simplify(H.det().subs(point))
        fxx = sp.simplify(H[0, 0].subs(point))
        value = function.subs(point)

        if D.is_positive and fxx.is_positive:
            results.append(f"Minimo en: f({point}, {value})")

        elif D.is_positive and fxx.is_negative: 
             results.append(f"Maximo en: f({point}, {value})")
        elif D.is_negative:
            results.append(f"Silla en: f({point}, {value})")
        else: 
            results.append(f"La prueba no es concluyente en: f({point}, {value})")
    return results
```

## c).Análisis algebraico de los Puntos Críticos en función de $a$

Para clasificar los puntos críticos obtenidos, aplicamos el **criterio de la segunda derivada para funciones de dos variables**. Recordando los resultados del inciso c, el determinante de la matriz Hessiana está dado por:

$$D(x, y) = 36xy - 9a^2$$

Y la segunda derivada parcial respecto a $x$ es:

$$f_{xx}(x, y) = 6x$$

Asumiendo que el parámetro $a \in \mathbb{R}$, analizamos el comportamiento del sistema dividiéndolo en dos escenarios principales:

---

### Caso 1: Si $a \neq 0$ (El parámetro es distinto de cero)

Bajo esta condición, cualquier número real elevado al cuadrado es estrictamente positivo ($a^2 > 0$). Evaluamos los dos puntos críticos hallados:

#### 1. Punto Crítico $P_1(0, 0)$
* **Evaluación del determinante:** Sustituyendo $x = 0$ e $y = 0$ en la expresión de $D$:
  $$D(0, 0) = 36(0)(0) - 9a^2 = -9a^2$$
* **Argumentación del signo:** Dado que $a^2 > 0$, al multiplicarlo por el coeficiente negativo $-9$, el determinante resulta estrictamente negativo para cualquier $a \neq 0$:
  $$D(0, 0) = -9a^2 < 0$$
* **Conclusión:** Al ser $D < 0$, el criterio de la segunda derivada establece de forma contundente que **el origen $(0, 0)$ es un punto de silla**.

#### 2. Punto Crítico $P_2(a, a)$
* **Evaluación del determinante:** Sustituyendo $x = a$ e $y = a$ en la expresión de $D$:
  $$D(a, a) = 36(a)(a) - 9a^2 = 36a^2 - 9a^2 = 27a^2$$
* **Argumentación del signo:** Como $a^2 > 0$, el producto por el coeficiente positivo $27$ garantiza que el determinante siempre será estrictamente mayor que cero:
  $$D(a, a) = 27a^2 > 0$$
  Esto confirma la existencia de un **extremo local**. Para determinar su naturaleza, analizamos el signo de $f_{xx}(a, a) = 6a$:
  
  * **Si $a > 0$:** La segunda derivada resulta positiva ($f_{xx} = 6a > 0$), lo que geométricamente implica una concavidad hacia arriba. Por lo tanto, el punto $(a, a)$ es un **mínimo local**.
  * **Si $a < 0$:** La segunda derivada resulta negativa ($f_{xx} = 6a < 0$), lo que geométricamente implica una concavidad hacia abajo. Por lo tanto, el punto $(a, a)$ es un **máximo local**.

---

### Caso 2: Caso especial si $a = 0$

Si el parámetro toma el valor exacto de cero, la función original se simplifica a $f(x, y) = x^3 + y^3$. El único punto crítico superviviente es el origen $(0,0)$.

* **Evaluación del determinante:** $$D(0, 0) = 36(0)(0) - 9(0)^2 = 0$$
* **Argumentación:** Cuando $D = 0$, el criterio de la segunda derivada **no es concluyente**. No obstante, si analizamos el entorno del origen de manera analítica:
  * Si nos aproximamos a lo largo del eje $x$ por la derecha ($x > 0, y = 0$), la función toma valores positivos: $f(x, 0) = x^3 > 0$.
  * Si nos aproximamos por la izquierda ($x < 0, y = 0$), la función toma valores negativos: $f(x, 0) = x^3 < 0$.
* **Conclusión:** Al existir tanto incrementos como decrementos del valor de la función en vecindades arbitrariamente cercanas al origen, se concluye que para $a = 0$, el punto $(0,0)$ también actúa como un **punto de silla**.

---

### Resumen General de Clasificación

| Punto Crítico | Condición del parámetro | Clasificación | Signo de $D$ | Signo de $f_{xx}$ |
| :--- | :--- | :--- | :--- | :--- |
| $(0,0)$ | $\forall \ a \in \mathbb{R}$ | **Punto de silla** | $D \leq 0$ | $0$ |
| $(a,a)$ | $a > 0$ | **Mínimo local** | $D > 0$ | $f_{xx} > 0$ |
| $(a,a)$ | $a < 0$ | **Máximo local** | $D > 0$ | $f_{xx} < 0$ |

Ejercicio 08

Se desea optimizar f sujeta a la curva de nivel g(x,y) = 0:
* f(x, y) = x² + y²
* g(x, y) = x² + xy + y² − 3 = 0


El método de multiplicadores de Lagrange plantea el sistema:
∇f = λ · ∇g y g(x, y) = 0

```python
# Exercise 08.
print('a) Declara x, y, λ como symbols y construye el sistema de 3 ecuaciones.')
# Symbols.
x, y, λ = sp.symbols('x y λ')
# Functions.
f = x**2 + y**2
g = x**2 + x*y + y**2 - 3
display(f)
display(g)
# Calculate derivatives.
f_fx = sp.diff(f, x)
f_fy = sp.diff(f, y)
g_fx = sp.diff(g, x)
g_fy = sp.diff(g, y)
# Equation system.
eq1 = sp.Eq(f_fx, λ * g_fx)
eq2 = sp.Eq(f_fy, λ * g_fy)
eq3 = sp.Eq(g, 0)

system = [eq1, eq2, eq3]
# Show system.
for i, eq in enumerate(system, 1):
    print(f"Ecuación {i}: {eq}")

print()
print('b) Resuelve el sistema completo con solve() obteniendo (x, y, λ).')

solutions = sp.solve(system, (x, y, λ ))
for solution in solutions:
    display(solution)
print()
print('c) Evalúa f en cada solución y determina cuál corresponde al mínimo y cuál al máximo.')
# 2. Listas para separar los resultados
maxim = []
minim = []

# 3. Clasificación usando condicionales
for sol in solutions:
    x_val, y_val, lam_val = sol
    f_val = f.subs({x: x_val, y: y_val})
    
    # En este problema sabemos que los valores extremos son 6 y 2
    if f_val == 6:
        maxim.append((x_val, y_val, lam_val))
    elif f_val == 2:
        minim.append((x_val, y_val, lam_val))

print()
print("Minimos:")
for x_v, y_v, lam in minim:
    print(f"Punto: ({x_v}, {y_v}) con λ = {lam}")

print()
print("Maximos:")
for x_v, y_v, lam in maxim:
    print(f"Punto: ({x_v}, {y_v}) con λ = {lam}")

print()
print('d) Verifica que cada solución satisface g(x, y) = 0 usando subs() y simplify().')

for i, sol in enumerate(solutions, 1):
    x_val, y_val, lam_val = sol
    
    
    g_eval = g.subs({x: x_val, y: y_val})
    
    g_simp = sp.simplify(g_eval)
    
    if g_simp == 0:
        print(f"Solución {i} (x={x_val}, y={y_val}):")
        print(f"  -> g({x_val}, {y_val}) = {g_simp}  [✓ OK]")
    else:
        print(f"Solución {i} (x={x_val}, y={y_val}):")
        print(f"  -> g({x_val}, {y_val}) = {g_simp}  [X NO OK]")
print()
print('e) Presenta un reporte final')

points_min = []
points_max = []

for sol in solutions:
    x_v, y_v, _ = sol
    f_val = f.subs({x: x_v, y: y_v})
    
    if f_val == 2:
        points_min.append((x_v, y_v))
    elif f_val == 6:
        points_max.append((x_v, y_v))

print("")
print("Mínimo de f sobre g=0:")
print(f"f{points_min[0]} = 2  en {points_min[0]}")
print(f"f{points_min[1]} = 2  en {points_min[1]}")

print("\nMáximo de f sobre g=0:")
print(f"f{points_max[0]} = 6  en {points_max[0]}")
print(f"f{points_max[1]} = 6  en {points_max[1]}")
```

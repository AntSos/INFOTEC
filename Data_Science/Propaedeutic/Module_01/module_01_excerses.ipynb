```python
"""
1. Hacer un programa en Python que resuelva la siguiente expresión aritmética
del área de un Triángulo, donde el usuario ingresa la base y la altura.

area = (base)(altura) / 2
"""
# Users input request.
while True:
    try:
        base = float(input("Ingresa el valor numérico de la base del triángulo: "))
        height = float(input("Ingresa el valor numérico de la altura del triángulo: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
        

# Triangle area calculation with a lambda function.
area = lambda b, h: (b * h) / 2
# Print area calculation result.
print(f"El área del triángulo con base {base} y altura {height} es: {area(base, height)}")
```

```python
"""
2. Hacer un programa en Python que resuelva la formula general para resolver ecuaciónes de 2o grado.
El usuario ingresa los valores de a, b y c.
"""
# Users input request.
while True:
    try:
        a = float(input("Ingresa el valor numérico de a: "))
        b = float(input("Ingresa el valor numérico de b: "))
        c = float(input("Ingresa el valor numérico de c: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# x1 and x2 calculation with lambdas functions.
x1 = lambda a, b, c: (-b + (((b**2.0) - 4.0 * (a * c))**0.5)) / (2.0 * a)
x2 = lambda a, b, c: (-b - (((b**2.0) - 4.0 * (a * c))**0.5)) / (2.0 * a)
# x1 and x2 calculation with users input.
try:
    result_x1 = x1(a, b, c)
    result_x2 = x2(a, b, c)
except ValueError as e:
    print(e, type(e))
# Print result.
print(f"Los dos valores que resuelven la ecuación con los parámetros: a = {a}, b = {b}, c = {c} son: X1 = {result_x1:.2f} y X2 = {result_x2:.2f}")
```

```python
"""
3. Hacer un programa en Python que calcule el salario semanal de un obrero, según los siguientes criterios:
- Si trabajó 40 horas o menos, se le paga $16 por hora.
- Si trabajó más de 40 horas, se le paga $16 por las primeras 40 horas, y $20.
por cada hora extra.
"""
# Users input request.
while True:
    try:
        hours = float(input("Ingrese el número de horas trabajadas: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Adjust payment according to worked hours.
if hours <= 40.0:
    payment = hours * 16.0
else: 
    payment = (40 * 16) + ((hours - 40) * 20)

# Print result
print(f"El salario a pagar por {hours:.0f} horas trabajadas es: ${payment:.2f}")
```

```python
"""
Hacer un programa en Python que clasifique el tipo de triángulo a partir de sus
lados ingresados:
- Equilátero: Los 3 son iguales.
- Isósceles: Tiene 2 lagos iguales y un lado distinto.
- Escaleno: Los 3 lados son distintos.
"""

# Users input request.
while True:
    try:
        equal_sydes = int(input("Ingrese el número de lados que son iguales en el triángulo: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Match a case.
match equal_sydes:
    case 3:
        print(f"Si presenta {equal_sydes} lados iguales es un triángulo equilátero.")
    case 2:
        print(f"Si presenta {equal_sydes} lados iguales es un triángulo isósceles.")
    case 0:
        print(f"Si ningún de sus lados son iguales es un triángulo escaleno.")
    case _:
        print(f"Con {equal_sydes} lados iguales no tienes ningún tipo de triángulo!.")
```

```python
"""
5. Hacer un programa en Python que calcule el Índice de Masa Corporal de una persona 
(IMC = peso[kg] / altura**2 [m]) e indique el estado en el que se encuentra esa persona 
en función del valor de IMC.

Valor de IMC  |  Diagnóstico
< 16          |  Criterio de ingreso en hospital
de 16 a 17    |  Infrapeso
de 17 a 18    |  Bajo peso
de 18 a 25    |  Peso normal (saludable)
de 25 a 30    |  Sobrepeso (obesidad de grado I)
de 30 a 35    |  Sobrepeso crónico (obesidad de grado II)
de 35 a 40    |  Obesidad premórbida (obesidad de grado III)
> 40          |  Obesidad mórbida (obesidad de grado IV)
"""
# Users input request.
while True:
    try:
        height = float(input("Ingrese su altura en metros: "))
        weight = float(input("Ingrese su peso actual en Kg: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Calculate IMC.
imc = weight / height**2
# Match a IMC value case with if conditionals.
if imc < 16:
    print(f"IMC = {imc:.2f} m/a2, criterio de ingreso en hospital.")
elif imc >= 16 and imc <= 17:
    print(f"IMC = {imc:.2f} m/a2, infrapeso.")
elif imc >= 17 and imc <= 18:
    print(f"IMC = {imc:.2f} m/a2, peso bajo.")
elif imc >= 18 and imc <= 25:
    print(f"IMC = {imc:.2f} m/a2, peso normal (saludable).")
elif imc >= 25 and imc <= 30:
    print(f"IMC = {imc:.2f} m/a2, sobrepeso(obesidad de grado I).")
elif imc >= 30 and imc <= 35:
    print(f"IMC = {imc:.2f} m/a2, sobrepeso crónico (obesidad de grado II).")
elif imc >= 35 and imc <= 40:
    print(f"IMC = {imc:.2f} m/a2, obesidad premórbida (obesidad de grado III)")
elif imc > 40:
    print(f"IMC = {imc:.2f} m/a2, obesidad mórbida (obesidad de grado IV)")

```

```python
"""
6. Hacer un programa en Python que, por un número entero dado por el usuario,
muestre la lista de números pares inferiores a él.
"""
# Users input request.
while True:
    try:
        number = int(input("Ingresa un numero y observa los numeros pares inferiores a este: "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Initialize an empty even numbers list
even_numbers = []
for n in range(number):
    if n % 2 == 0:
        if n == 0:
            continue
        else:
            even_numbers.append(n)
# Sort the list in reverse
even_numbers.sort(reverse=True)
# Print the result
print(f" Los valores pares por debajo del número {number} son: ")
for e in even_numbers:
    print(e)
```

```python
"""
8. Hacer un programa en Python que imprima la serie de la sumatoria Σ 1/n, donde n es ingresado por el usuario.
a. Ejemplo de salida: 1/1, 1/2, 1/3, … , 1/n
"""
# Users input request.
while True:
    try:
        number = int(input("Ingresa un numero y obten la suma de 1 /n : "))
        break
    except ValueError as e:
        print(e, type(e))
        print("Debes ingresar solo valores numéricos.")
# Initialize an empty list to store the values.
sum_values = []
# Create and print the division values.
for n in range(1, number + 1):
    print(f"{1} / {n}")
    sum_values.append(1 / n)
# Print the sum total value.
print(f"La suma de estos valores es: {sum(sum_values):.2f}")
```

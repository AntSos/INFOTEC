```python
import numpy as np
import sympy as sp
```

```python
# Slicing and comparing.
A = np.array([[2, 4, 6], [8, 0, 1]])
B = np.array([[-(-2), 16/4, 6], [8, 0, -3/-3]])

print(np.shape(A))
print(A[0][1], A[1][2])
print(np.equal(A, B))
```

```python
# Adittion and substraction.
C = np.array([[1, 8], [11, 4], [0, 9]])
D = np.array([[6, -7], [3, 15], [2, 5]])

print(C + D)
print(C - D)
```

```python
# Addition, substraction and multiplication.
E = np.array([[-1, 2, 6], [8, -7, 5]])
F = np.array([[-3, -5, -2], [0, 4, 7]])

print((4*A) + (3*B), end='\n\n')
print((5*A) - (2*B), end='\n\n')
```

```python
# Transponse.
G = np.array([[2, 0, -1], [4, 6, 8]])
print(np.transpose(G), end='\n\n')
print(G.T, end='\n\n')
# double transponse.
print(G.T.T, end='\n\n')
```

```python
# 
H = np.array([[1, 2, 3], [4, 5, 6]])
I = np.array([[7, 8, 9], [10, 11, 12]])

print((H + I).T, end='\n\n')
print((H.T + I.T), end='\n\n')
```

```python
# Column-row vectors multiplication.
row = np.array([[-2, 4, 6]])
col = np.array([[-3], [-6], [9]])

print(col @ row)
```

```python
# 3X2, 2X3 multiplication matrix.
J = np.array([[2, 4], [1, 3], [5, 8]])
K = np.array([[0, 6, -1], [1, 7, 11]])

print(J @ K)
```

```python
L = np.array([[2, -3, 4], [-7, 8, 1]])
M = np.array([[2, -9], [-4, 5], [7, -8]])
N = np.array([[-6, 3], [5, 7]])

print((L @ M) @ N, end='\n\n')
print(L @ (M @ N), end='\n\n')
```

```python
# Equation system solving.
P = np.array([[1,1], [2, 1]])
q = np.array([[5], [7]])


R = np.array([[2, 10, -40], [-4, -7, 41], [1, 3, -12]])
s = np.array([[-4], [-31], [-6]])


print(np.linalg.solve(P, q), end='\n\n')
print(np.linalg.solve(R, s), end='\n\n')
```

```python
x, y, z, a, b, c = sp.symbols('x y z a b c')

eq1 = sp.Eq(x + y, a)
eq2 = sp.Eq(x - y, b)
eq3 = sp.Eq(y, c)

sol = sp.solve((eq1, eq2, eq3), (x, y, a, b, c))

for k, v in sol.items():
    print(f"{k} = {v}")
```

```python
eq4 = sp.Eq(-5*x -4*y + 10*z, a)
eq5 = sp.Eq(-8*x -8*y - 5*z, b)

sol2 = sp.solve((eq4, eq5), (x, y, z, a, b))

for k, v in sol2.items():
    print(f"{k} = {v}")
```

```python
T = np.array([[1, -2], [-3, 7]])
print(np.linalg.det(T))
O = np.array([[2, 0], [1, -1]])
print(np.linalg.det(O))
```

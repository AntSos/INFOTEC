```python
import sympy as sp
import numpy as np
```

```python
def sp_system_sol(eqs, symbols):
    """ A simple funtion to solve a system of equations with sympy"""
    # Convert system of equations to a matrix A and vector b.
    A, b = sp.linear_eq_to_matrix(eqs, symbols)
    # System number of rows and columns string.
    system = f"Sistema: {A.shape[0]}X{A.shape[1]}"
    # Solution to the system.
    solution = sp.linsolve((A, b))
    # System withot solutions.
    if solution == sp.EmptySet:
        txt_solution = "El sistema no tiene solucion"
        l_sol = None
        return system, l_sol, txt_solution
    # System with one or many solutions.
    else:
        l_sol = list(solution)[0]
        solutions = set()
        for e in l_sol:
            solutions.update(e.free_symbols)

        if solutions:
            txt_solution = "El sistema tiene multiples soluciones"
        else:
            txt_solution = "El sistema tiene solucion unica"
    
    return system, l_sol, txt_solution
```

```python
def np_system_solve(eqs, symbols):
    """ A simple funtion to solve a square and non-square system of equations with sympy and numpy"""
    # Convert system of equations to a matrix A and vector b.
    A, b = sp.linear_eq_to_matrix(eqs, symbols)
    # Convert sympy to numpy.
    np_A = np.array(A).astype(np.float64)
    np_b = np.array(b).astype(np.float64)
    np_solution = None
    # Check the if the matrix is not square.
    if A.shape[0] != A.shape[1]:
        # Solve non-square system with numpy.
        try:
            np_solution, residuals, rank, s = np.linalg.lstsq(np_A, np_b, rcond=None)
            return np_solution
        except np.linalg.LinAlgError as e:
            return "numpy error", e
    else:
        # Solve the square system with numpy.
        try:
            np_solution = np.linalg.solve(np_A, np_b)
            return np_solution
        except np.linalg.LinAlgError as e:
            return "numpy error", e
```

```python
# sympy numpy way to solve the system.
x, y = sp.symbols('x y')
# First set of equations.
eqs1 = [sp.Eq(2*x + 3*y, 8),
        sp.Eq(x - 4*y, -5)
       ] 
symbols = [x, y]
# Use sp_system_sol() function.
system, l_sol, txt_solution = sp_system_sol(eqs1, symbols)
# Use np_system_solve function.
np_solution = np_system_solve(eqs1, symbols)
# Check function returned values and ptint results.
if txt_solution == "El sistema tiene solucion unica" or txt_solution == "El sistema tiene multiples soluciones":
    print(system)
    print(txt_solution)
    
    # Check np returned values.
    if np_solution is None or isinstance(np_solution, tuple):
        print(np_solution)
    else:
        print(f"Numpy: x = {np_solution[0][0]:.3f}, y = {np_solution[1][0]:.3f}")
    
    if txt_solution == "El sistema tiene solucion unica":
        print(f"Sympy: x = {l_sol[0]}, y = {l_sol[1]}")
    else:
        print(f"Sympy solution: {l_sol}")
else:
    print(txt_solution)
```
